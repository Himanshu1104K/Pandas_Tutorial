Four methods to Do the above action more powerful way
1. apply
2. applymap
3. Map 

```python
def update_emails(emails):
    return emails.upper()
```

```python
df.columns = df.columns.str.lower()
```

```python
df['emails']=df['emails'].apply(update_emails)
```

```python
df
```

```python
# df['emails'] = [x.lower() for x in df.emails]
df['emails'] = df['emails'].apply(lambda x : x.upper())
```
